# Foundry datasets: JSON to versioned JSONL


```bash
cd "$(git rev-parse --show-toplevel)" &&
PYTHON="$PWD/.venv/bin/python" &&
FOUNDRY_PROJECT_ENDPOINT="https://ai-projplan-axowf3.services.ai.azure.com/api/projects/project-planning" &&
PUBLISH_DATASETS=no &&
printf "Dataset publication: %s (set PUBLISH_DATASETS=yes to authorize upload)\n" "$PUBLISH_DATASETS"
```

Generate every configured JSONL file and verify that it matches its JSON sources. Commit both JSON and JSONL changes. This cell performs no cloud writes.

```bash
"$PYTHON" scripts/foundry_datasets.py generate &&
"$PYTHON" scripts/foundry_datasets.py check
```

Optional upload: sign in with Azure CLI, then explicitly set PUBLISH_DATASETS=yes in the first cell before rerunning it. The next cell uses the same upload command as GitHub Actions. It creates content-hash dataset versions, not agent versions or evaluation runs. The local release manifest is ignored by Git; these notebook outputs remain versioned.

```bash
case "$PUBLISH_DATASETS" in
  yes) "$PYTHON" scripts/foundry_datasets.py publish \
    --endpoint "$FOUNDRY_PROJECT_ENDPOINT" --confirm \
    --manifest evaluations/notebook_runs/dataset-publication/release.json ;;
  no) printf "Upload skipped: PUBLISH_DATASETS=no. No Foundry resources changed.\n" ;;
  *) printf "PUBLISH_DATASETS must be yes or no.\n" >&2; false ;;
esac
```
